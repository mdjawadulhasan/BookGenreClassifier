# Data scraping

### Import the scraping libraries

```python
from bs4 import BeautifulSoup
import requests
```

### Scraping Fiction Books

```python
title=[]
category=[]
book_url=[]	
image_url=[]

for i in range(1,5):
    fiction_category_link=f"https://books.toscrape.com/catalogue/category/books/fiction_10/page-{i}.html"
    response=requests.get(fiction_category_link)
    soup=BeautifulSoup(response.content,'html.parser')
    fiction_books=soup.find_all('article',class_='product_pod')
    
    for book in fiction_books:
        fiction_book_title=book.find('h3').find('a')['title']
        fiction_book_url= book.find('a')['href'].replace('../../../', '')
        fiction_image_url= book.find('img')['src'].replace('../../../../', '')
        
        print('Scapped Book:'+fiction_book_title)

        title.append(fiction_book_title)
        book_url.append("https://books.toscrape.com/catalogue/" + fiction_book_url)
        image_url.append("https://books.toscrape.com/" + fiction_image_url)
        category.append('Fiction')
```

```python
print('Total Fiction Books Scrapped: ' + str(len(title)))
```

### Scraping NonFiction Books

```python
for i in range(1,7):
    non_fiction_category_link=f"https://books.toscrape.com/catalogue/category/books/nonfiction_13/page-{i}.html"
    response=requests.get(non_fiction_category_link)
    soup=BeautifulSoup(response.content,'html.parser')
    non_fiction_books=soup.find_all('article',class_='product_pod')
    
    for book in non_fiction_books:
        non_fiction_book_title=book.find('h3').find('a')['title']
        non_fiction_book_url= book.find('a')['href'].replace('../../../', '')
        non_fiction_image_url= book.find('img')['src'].replace('../../../../', '')
        
        print('Scapped Book:'+non_fiction_book_title)

        title.append(non_fiction_book_title)
        book_url.append("https://books.toscrape.com/catalogue/" + non_fiction_book_url)
        image_url.append("https://books.toscrape.com/" + non_fiction_image_url)
        category.append('Nonfiction')
```

```python
print('Total Books Scrapped: ' + str(len(title))) 
```

### Add them to Dataframe

```python
import pandas as pd

df = pd.DataFrame({
    'title': title,
    'category': category,
    'book_url': book_url,
    'image_url': image_url
})
```

```python
df.head(5)
```

### Collect Book Descriptions

```python
description = []

for book_description_link in df['book_url']:
    response=requests.get(book_description_link)
    soup=BeautifulSoup(response.content,'html.parser')
    book_description=soup.find('div', id='product_description')
    description.append(book_description.find_next('p').text)   
```

```python
print('Total Book Descriptions Scrapped: ' + str(len(description)))
```

```python
df['description'] = description
```

```python
df.head(2)
```

### Collect Images  

```python
image_path = []

for i in range(0,175):
    response=requests.get(df['image_url'][i])
    path=f'images/book_{i}.jpg'
    with open(path, 'wb') as f:
        f.write(response.content)
    image_path.append(path)
```

```python
print('Total Image Scrapped: ' + str(len(image_path)))
```

```python
df['image_path'] = image_path
```

```python
df.head()
```

### Store Data in CSV

```python
df.to_csv("data/raw/books_raw.csv", index=False)
```
