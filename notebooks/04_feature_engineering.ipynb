# Feature Engineering

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/cleaned/books_cleaned.csv')
```

## combine title and description and create a new column

```python
df["title_description"] = ( "Title: " + df["title"].fillna("").astype(str)+ " Description: " + df["description"].fillna("").astype(str))
```

```python
print(df[["title", "description", "title_description"]].head())
```

## save the changes

```python
df.to_csv("../data/cleaned/books_cleaned.csv", index=False)
```
